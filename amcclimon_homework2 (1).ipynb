### Question 1
Load the gun violence data set (gun-violence-data_01-2013_03-2018.csv) into a
DataFrame.

Question 1a. Use the appropriate command to report the data type for each column

```python
# load gun violence data
gun_violence = pd.read_csv("gun-violence-data_01-2013_03-2018.csv", low_memory=False)
```

```python
# report data types
print("Gun violence column data types:")
print(gun_violence.dtypes)
```

Question 1b. Filter the original data frame to show incidents that occurred in Iowa City, IA. Be
sure to not include incidents that occurred in Iowa City, CA. Save results to a new
data frame named iowa_city_ia. How many incidents occurred in Iowa City,
IA? Show the new data frame.

```python
# save incidents in Iowa City and report count/display 
iowa_city_ia = gun_violence[(gun_violence["state"] == "Iowa") & (gun_violence["city_or_county"] == "Iowa City")].copy()
print("Number of incidents in Iowa City, IA:", len(iowa_city_ia))
display(iowa_city_ia)
```

Question 1c. Filter the original data frame to show incidents that occurred in both Iowa City
(like step b) and Cedar Rapids, IA. Be sure not to include Cedar Rapids, NE. Save
results to a new data frame named corridor_ia. How many incidents
occurred in Iowa City and Cedar Rapids, IA? Show the new data frame

```python
# save incidents in IC and CR to corridor_ia, show df
corridor_ia = gun_violence[(gun_violence["state"] == "Iowa") & (gun_violence["city_or_county"].isin(["Iowa City", "Cedar Rapids"]))].copy()
print("Number of incidents in Iowa City and Cedar Rapids, IA:", len(corridor_ia))
display(corridor_ia)
```

Question 1d. Save all Iowa incidents into a new data frame named iowa_df and save it as a
JSON file named iowa_incidents.json. Verify the created JSON file is valid
using https://jsonlint.com. Do not submit a copy of your JSON file, just the
working python code to create the file. We will validate it during grading.

```python
# save all IA incidents in iowa_df and export iowa_incidents.json
iowa_df = gun_violence[gun_violence["state"] == "Iowa"].copy()
iowa_df.to_json("iowa_incidents.json", orient="records", indent=4)

# reopen the export to check its JSON syntax.
with open("iowa_incidents.json", "r", encoding="utf-8") as json_file:
    iowa_incidents = json.load(json_file)
print("Created valid JSON file: iowa_incidents.json")
print("Number of Iowa incidents exported:", len(iowa_incidents))
```

Question 1e. In 2016, which cities and states had the most fatalities? Display the maximum
number killed in 2016. Show n_killed, city_or_county, and state columns for
records that match the maximum number killed.

```python
# parse dates separately to preserve the original df
incident_dates = pd.to_datetime(gun_violence["date"])
incidents_2016 = gun_violence[incident_dates.dt.year == 2016]
city_fatalities = incidents_2016.groupby(
    ["city_or_county", "state"], as_index=False
)["n_killed"].sum()
max_killed_2016 = city_fatalities["n_killed"].max()
print("max fatalities 2016:", max_killed_2016)
display(city_fatalities.loc[
    city_fatalities["n_killed"] == max_killed_2016,
    ["n_killed", "city_or_county", "state"]
])
```

### Question 2
Load the FIFA 21 complete player data set (players_21.csv) into a data frame named fifa_21.

```python
# load data in fifa_21
fifa_21 = pd.read_csv("players_21.csv", low_memory=False)
```

Question 2a. Subset the following player IDs (sofifa_id) into a separate data frame named
selected_players. Save only the player id and long name fields into this new
data frame.

```python
# subset ids into selected_players and display
player_ids = [231747, 153079, 162835, 200104, 223848, 208421]
selected_players = fifa_21.loc[fifa_21["sofifa_id"].isin(player_ids), ["sofifa_id", "long_name"]].copy()
display(selected_players)
```

Question 2b. Subset players with a height greater than or equal to 191cm and a weight greater
than or equal to 90kg into a separate data frame named big_players. Display
the number of players in this new data frame.

```python
# save players at least 191 cm and 90 kg in big_players 
big_players = fifa_21[(fifa_21["height_cm"] >= 191) & (fifa_21["weight_kg"] >= 90)].copy()
print("Players at least 191 cm tall and 90 kg:", len(big_players))
```

Question 2c. What is the age of the oldest player(s)? The youngest? What is the median age
and the mean age of all players. Store each in variables and display them. Display
results in whole numbers.

```python
# store and display the oldest, youngest, median, and mean ages then display
oldest_age = fifa_21["age"].max()
youngest_age = fifa_21["age"].min()
median_age = fifa_21["age"].median()
mean_age = fifa_21["age"].mean()
print(f"Oldest player age: {oldest_age:.0f} years")
print(f"Youngest player age: {youngest_age:.0f} years")
print(f"Median player age: {median_age:.0f} years")
print(f"Mean player age: {mean_age:.0f} years")
```

Question 2d. Show the long name, age, and height for the oldest player(s) who prefer their left
foot

```python
# show long name, age, and height for the oldest players who are lefties
left_footed_players = fifa_21[fifa_21["preferred_foot"] == "Left"].copy()
oldest_left_age = left_footed_players["age"].max()
oldest_left_players = left_footed_players.loc[left_footed_players["age"] == oldest_left_age, ["long_name", "age", "height_cm"]].copy()
print("Oldest left-footed players (height in cm):")
display(oldest_left_players)
```
